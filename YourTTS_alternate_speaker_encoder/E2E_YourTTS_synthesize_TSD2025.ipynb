# Synthesis

```python
%%bash

#------
# function: synthesize speech using a reference wav
#------
run_tts_refwav () {

    speaker="${speakerID}_${dur}s"
    ref_wav="/storage/plzen4-ntis/projects/public/mkunes/YourTTS/CRo_recnici_na_poslechovy_test/spojene_nahodne_naTSD2025/${speaker}.wav"
    
    ## ECAPA-TDNN, trained with SCL
    main_dir="/media/mkunes/Data/Experiments/YourTTS/TSD2025/models/"
    subdir="SPT-MGW.JMa_ECAPA-TDNN-VoxCeleb_alwaysSCL"
    checkpoint="checkpoint_81.pth"
    config="config.json"
    
    tts_run_one_model
}


#------
# function: do the actual speech synthesis
#------
tts_run_one_model () {

    model_dir="${main_dir}/${subdir}"
    outdir="/media/mkunes/Data/Experiments/YourTTS/TSD2025/tts_out/${subdir}/${checkpoint}_ref-wav"
    mkdir -p "$outdir"
    
    tts --text "$text" \
        --config_path $model_dir/$config \
        --model_path $model_dir/$checkpoint \
        --out_path "$outdir/${speaker}_1ref_${textID}.wav" \
        --speaker_wav $ref_wav
}

#===== Speakers =======
dur=30

text="$ tuto vjetu nejspIS Jigdi neReknu, # ale bila vitvoRena pomocI sintEzi ReCi na katedRe kibernetiki. $"
textID="nikdyNereknu"
for speakerID in "Abrahámová-Jitka" "Bílá-Lucie" "Galaš-Radek" "Laurinová-Sabina" "Němcová-Miroslava" "Rázlová-Regina" "Anderle-Jan" "Čunek-Jiří" "Grund-Jiří" "Magdoňová-Jana" "Preiss-Viktor" "Skopeček-Jan" "Babiš-Andrej" "Drahoš-Jiří" "Kostková-Tereza" "Mazalová-Blanka" "Procházka-Martin" "Výborná-Lucie" "Baxa-Martin" "Dvořák-Libor" "Křížová-Markéta" "Molavcová-Jitka" "Rakušanová-Lída" "Zeman-Miloš"
do
    run_tts_refwav
done





```
